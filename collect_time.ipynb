```python
import os.path as osp
import pandas as pd
from datetime import timedelta

folder = "dataset/mossbauer/dft_ms"

test_set = pd.read_csv('test_set.tsv', sep='\t')
test_set.index = test_set['FILE']

dft_ms_times = []

for file in test_set['FILE']:
    with open(osp.join(folder, f'{file[:-4]}.log'),'r') as f:
        lines = f.readlines()
        time_line = 0
        for i in lines:
            if i[:14] == ' Job cpu time:':
                break
            time_line += 1

        line_split = lines[time_line].split()
        
    new_time = timedelta(days=float(line_split[3]), 
                         hours=float(line_split[5]), 
                         minutes=float(line_split[7]), 
                         seconds=float(line_split[9]))
    dft_ms_times.append(new_time)

```

```python
import os.path as osp
import pandas as pd
from datetime import timedelta

folder = "predictor/time_test/out"

test_set = pd.read_csv('test_set.tsv', sep='\t')
test_set.index = test_set['FILE']

gfn1_opt_times = []

for file in test_set['FILE']:
    with open(osp.join(folder, file[:-4], 'xtb.out'),'r') as f:
        lines = f.readlines()
        time_line = 0
        for i in lines:
            if i[:13] == ' *  cpu-time:':
                break
            time_line += 1

        line_split = lines[time_line].split()
        
    new_time = timedelta(days=float(line_split[2]), 
                         hours=float(line_split[4]), 
                         minutes=float(line_split[6]), 
                         seconds=float(line_split[8]))
    gfn1_opt_times.append(new_time)
```

```python
low_level_dft_times = [x + y for x, y in zip(dft_ms_times, gfn1_opt_times)]

print(f"Total: {sum(low_level_dft_times, start=timedelta())}")
print(f"Average: {sum(low_level_dft_times, start=timedelta())/len(test_set.index)}")
print(f"Min: {min(low_level_dft_times)}")
print(f"Max: {max(low_level_dft_times)}")
print(f"Range: {max(low_level_dft_times) - min(low_level_dft_times)}")
```

```python
import os.path as osp
import pandas as pd
from datetime import timedelta

folder = "predictor/time_test/out"

test_set = pd.read_csv('test_set.tsv', sep='\t')
test_set.index = test_set['FILE']

gfn1_sp_times = []

for file in test_set['FILE']:
    with open(osp.join(folder, file[:-4], 'tblite.out'),'r') as f:
        lines = f.readlines()
        time_line = 0
        for i in lines:
            if i[:7] == ' total:':
                break
            time_line += 1

        line_split = lines[time_line].split()
        
    if('min' not in line_split): 
        new_time = timedelta(seconds=float(line_split[1]))
    else:
        new_time = timedelta(minutes=float(line_split[1]),
                             seconds=float(line_split[3]))
    gfn1_sp_times.append(new_time)
```

```python
ML_prediction_time = timedelta(seconds=0.8106993548572063)

gfn1_ml_prediction_times = [x + y + ML_prediction_time/len(test_set.index) for x, y in zip(gfn1_opt_times, gfn1_sp_times)]

print(f"Total: {sum(gfn1_ml_prediction_times, start=timedelta())}")
print(f"Average: {sum(gfn1_ml_prediction_times, start=timedelta())/len(test_set.index)}")
print(f"Min: {min(gfn1_ml_prediction_times)}")
print(f"Max: {max(gfn1_ml_prediction_times)}")
print(f"Range: {max(gfn1_ml_prediction_times) - min(gfn1_ml_prediction_times)}")
```
